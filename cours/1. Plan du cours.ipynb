## Partie 1 — Fondamentaux de la Data Science

* Introduction à la Data Science et au Machine Learning.
* Environnement Python : Jupyter Notebook, Miniconda, VS Code.
* Bibliothèques : NumPy, Pandas, Matplotlib et Seaborn.
* Chargement de fichiers CSV, Excel et JSON.
* Exploration et visualisation des données.
* Mini-projet : explorer un jeu de données de ventes.

## Partie 2 — Préparation des données tabulaires

* Nettoyage des données manquantes et des doublons.
* Détection et traitement des valeurs aberrantes.
* Encodage des variables catégorielles.
* Normalisation et standardisation.
* Séparation des données en ensembles d'entraînement et de test.
* Pipelines de prétraitement avec Scikit-learn.
* Mini-projet : préparer un dataset de clients avant l'entraînement d'un modèle.

## Partie 3 — Machine Learning supervisé

* Données étiquetées : on connaît déjà les réponses attendues pendant l'entraînement.
* Régression : Linear Regression, Decision Tree Regressor, Random Forest Regressor.
* Classification : Logistic Regression, KNN, Decision Tree, Random Forest, SVM.
* Méthodes d'ensemble : Bagging, Boosting, XGBoost ou équivalent.
* Évaluation : accuracy, precision, recall, F1-score, MAE et RMSE.
* Surapprentissage, sous-apprentissage et validation croisée.
* Mini-projet : prédire le prix d'un logement ou le départ d'un client.

## Partie 4 — Machine Learning non supervisé

* Clustering : K-Means, DBSCAN et clustering hiérarchique.
* Réduction de dimension : PCA, utilisée ici de manière intuitive.
* Détection d'anomalies.
* Segmentation de clients.
* Visualisation des groupes et interprétation des résultats.
* Mini-projet : segmenter des clients selon leurs habitudes d'achat.

## Partie 5 — NLP : Natural Language Processing

* Nettoyage et préparation des textes.
* Tokenisation, stop words, stemming et lemmatisation.
* Représentation des textes : Bag of Words et TF-IDF.
* Classification de textes et analyse de sentiment.
* Reconnaissance d'entités nommées (NER).
* Embeddings, modèles préentraînés et introduction aux Transformers.
* Mini-projet : analyser automatiquement les avis des clients d'un hôtel.

## Partie 6 — Séries temporelles

* Manipulation des dates et index temporels avec Pandas.
* Analyse des ventes, du trafic ou de la consommation.
* Construction de variables à partir du passé.
* Prévision avec des modèles simples de Machine Learning.
* Mini-projet : prévoir les ventes du mois suivant.
