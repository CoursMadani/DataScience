## 1.1. Qu'est-ce que la Data Science ?

La ***Data Science***, ou science des données, est un domaine qui vise à _extraire des informations_ utiles à partir de données, afin de _comprendre une situation_, _répondre à une question_ ou _aider à prendre une décision_.

<center><img src="../images/ds3.png" width="800"></center>

La Data Science est un domaine pluridisciplinaire qui se situe au croisement de plusieurs disciplines complémentaires : 
* **L'Informatique** pour collecter, stocker et traiter de grandes quantités de données grâce à la programmation, aux algorithmes et aux bases de données.
* **Les mathématiques et les statistiques** pour analyser les données, identifier des tendances, construire des modèles et mesurer l’incertitude.
* **L’intelligence artificielle (IA)**, notamment le Machine Learning et le Deep Learning, permet d’apprendre à partir des données afin de réaliser des prédictions et d’automatiser certaines tâches.
* **La visualisation** des données facilite l’interprétation des résultats et leur communication à travers des graphiques, des tableaux de bord et des rapports.
* Enfin, **les domaines d’application**, tels que la santé, la finance, l’industrie et l’éducation, apportent les connaissances métier nécessaires pour interpréter les résultats et prendre des décisions pertinentes.

Ainsi, le chevauchement de ces disciplines permet à la Data Science de transformer des données brutes en connaissances exploitables pour comprendre les phénomènes, anticiper les évolutions et aider à la prise de décision.

<img src="../images/ds2.png" width=700>

**La data science combine principalement** :

la compréhension du problème métier ;

* la collecte et la préparation des données ;
* la programmation ;
* l'analyse et la visualisation ;
* le Machine Learning, lorsque cela est pertinent ;
* l'interprétation et la communication des résultats.

**Exemple concret** : Imagine qu'un hôtel possède les données suivantes :

| Client | Nombre de nuits | Dépenses | Note donnée |
| --- | --- | --- | --- |
| A | 2 | 1 500 DH | 4/5 |
| B | 5 | 4 800 DH | 5/5 |
| C | 1 | 600 DH | 2/5 |
| D | 3 | 2 200 DH | 4/5 |
| E | 4 | 3 500 DH | 5/5 |


À partir de ces données, on peut répondre à différentes questions :

* Quel est le montant moyen dépensé par client ?
* Quels clients dépensent le plus ?
* Existe-t-il une relation entre la durée du séjour et les dépenses ?
* Peut-on identifier des groupes de clients ayant des comportements similaires ?
* Peut-on prédire la note qu'un nouveau client pourrait donner ?

Ces questions correspondent à différents types de travaux en Data Science. Certaines ne nécessitent que des calculs et des graphiques ; d'autres peuvent nécessiter du Machine Learning.

**À retenir** : la Data Science ne consiste pas uniquement à entraîner des modèles. Il faut d'abord comprendre les données et la question à résoudre.

**Remarque** : Les termes Data Analytics, Data Science, Machine Learning et Deep Learning sont proches, mais ils ne désignent pas exactement la même chose.<br>
- ***Data Analytics*** : Analyse les données existantes pour comprendre ce qui s'est passé ou ce qui se passe. Par exemple, calculer le chiffre d'affaires mensuel d'un hôtel.
- ***Data Science*** : Utilise différentes méthodes d'analyse et de modélisation pour extraire des informations utiles. Par exemple, étudier les facteurs associés aux annulations de réservation.
- ***Machine Learning (ML)*** : Permet à un système d'apprendre des régularités à partir d'exemples pour effectuer des prédictions ou découvrir des structures. Par exemple, prédire si une réservation risque d'être annulée.
- ***Deep Learning*** : Sous-domaine du Machine Learning qui utilise des réseaux de neurones comportant plusieurs couches. Par exemple, analyser des images ou traiter automatiquement des textes complexes.

Ces domaines se chevauchent. Un projet de Data Science peut utiliser le Machine Learning et le Deep Learning, mais il peut également se limiter à une analyse descriptive.

## 1.2. Les principales étapes d'un projet de Data Science

Un projet suit généralement une succession d'étapes.

* 01 - Définir le problème : Quelle question veut-on résoudre ?
* 02 - Collecter les données : CSV, Excel, base de données, API…
* 03 - Explorer les données : Comprendre les colonnes, les types et la qualité.
* 04 - Préparer les données : Corriger les incohérences et traiter les valeurs manquantes.
* 05 - Analyser ou entraîner un modèle : Appliquer les méthodes adaptées à la question.
* 06 - Évaluer les résultats : Vérifier la qualité et les limites du résultat.
* 07 - Communiquer ou déployer : Présenter les conclusions ou intégrer le modèle à une application.

Le processus est souvent itératif : si les résultats ne sont pas satisfaisants, on peut revenir à la préparation, aux données ou à la définition du problème.

Exemple illustratif : Un hôtel souhaite réduire les annulations.

* Problème : identifier les réservations présentant un risque d'annulation.
* Données : historique des réservations, délais entre réservation et arrivée, type de chambre, canal de réservation.
* Exploration : vérifier les valeurs manquantes et les annulations historiques.
* Préparation : traiter les catégories et les données manquantes.
* Modélisation : entraîner un classifieur.
* Évaluation : mesurer sa capacité à détecter les annulations sur des données non utilisées pour l'entraînement.
* Utilisation : aider le personnel à décider quelles réservations surveiller.

Il faut surtout éviter d'utiliser des informations qui ne seraient connues qu'après l'annulation : elles rendraient la prédiction artificiellement performante.

## 1.3. Les types de données

Avant d'utiliser un algorithme, il faut savoir quelle information contient chaque colonne.

| Type | Exemple | Utilisation possible |
| --- | --- | --- |
| Numérique | Prix, âge, durée | Régression, calculs, analyse |
| Catégoriel nominal | Ville, pays, canal | Classification, segmentation |
| Catégoriel ordinal | Faible, moyen, élevé | Classement, prédiction |
| Date et heure | Date de réservation | Analyse temporelle, prévision |
| Texte | Avis client | NLP, analyse de sentiment |
| Identifiant | Numéro client | Rapprochement des données |
| Cible | Annulation : oui/non | Variable à prédire |


Un identifiant n'est pas automatiquement une bonne variable prédictive. Par exemple, le numéro d'un client ne possède pas nécessairement de signification métier exploitable.
